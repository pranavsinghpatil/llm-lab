# Character Tokenizer Walkthrough & Visualization

This notebook demonstrates how the `CharacterTokenizer` and `Vocabulary` classes process text step-by-step.

````python
import sys
import os
# Ensure project root is in path
sys.path.append(os.path.abspath('../../'))

from tokenizer.implementation import CharacterTokenizer
````

## Step 1: Initialize and Fit Tokenizer

````python
text = "banana"
tokenizer = CharacterTokenizer()
tokenizer.fit(text)

print(f"Vocabulary size: {tokenizer.vocab_size}")
print(f"Token to ID Mapping: {tokenizer.vocab.token_to_id}")
````

## Step 2: Encode String to Token IDs

```
"banana" ---> ['b', 'a', 'n', 'a', 'n', 'a'] ---> [2, 1, 3, 1, 3, 1]
```

````python
tokens = tokenizer.encode(text)
print(f"Raw Text : {text}")
print(f"Characters: {list(text)}")
print(f"Token IDs : {tokens}")
````

## Step 3: Decode Token IDs back to String

```
[2, 1, 3, 1, 3, 1] ---> "banana"
```

````python
decoded = tokenizer.decode(tokens)
print(f"Decoded String: '{decoded}'")
assert decoded == text, "Decoded string must match original!"
````

## Step 4: Unknown Character Handling (<UNK>)

Testing `"bandana"` when fitted only on `"banana"` (`'d'` is unseen).

````python
unknown_text = "bandana"
unknown_tokens = tokenizer.encode(unknown_text)
unknown_decoded = tokenizer.decode(unknown_tokens)

print(f"Input Text     : {unknown_text}")
print(f"Encoded Tokens : {unknown_tokens}")
print(f"Decoded Output : {unknown_decoded}")
````
